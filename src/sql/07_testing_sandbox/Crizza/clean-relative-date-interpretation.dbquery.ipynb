```sql
%sql
-- The date column stores RELATIVE days from course start (not calendar dates)
-- Negative = days before course start, Positive = days after course start

SELECT 
  date as relative_day,
  CASE 
    WHEN date < 0 THEN CONCAT(ABS(date), ' days BEFORE course start')
    WHEN date = 0 THEN 'Course START day'
    ELSE CONCAT(date, ' days AFTER course start')
  END as interpretation,
  COUNT(*) as total_interactions,
  COUNT(DISTINCT id_student) as unique_students,
  AVG(sum_click) as avg_clicks_per_interaction
FROM oulad.clean.student_vle
WHERE date IN (-25, -20, -10, 0, 1, 30, 100, 200, 250, 269)
GROUP BY date
ORDER BY date;
```
